TREATING IMBALANCE USING SMOTE

```python
#  IMPORTS
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import (
    classification_report,
    confusion_matrix,
    precision_recall_curve,
    auc,
    f1_score
)
from imblearn.over_sampling import SMOTE
import warnings
warnings.filterwarnings("ignore")

# SPLIT FIRST — keep test set REALISTIC (imbalanced)
print("Step 1: Splitting data (preserving original imbalance in test set)...")
X_train, X_test, y_train, y_test = train_test_split(
    X, y, 
    test_size=0.2, 
    stratify=y,        # ensures train/test have same % of class 1
    random_state=42
)

print("Original training set distribution:")
print(y_train.value_counts().sort_index())
print("\nOriginal test set distribution:")
print(y_test.value_counts().sort_index())
```

```python
# WE ARE GOING TO APPLY SMOTE ONLY TO THE TRAINING DATA
print("\n Step 2: Applying SMOTE to training data only...") 
smote = SMOTE(random_state=42)
X_train, y_train = smote.fit_resample(X_train, y_train)
print("After SMOTE — training set distribution:")
print(y_train.value_counts().sort_index())
# Now the training set is balanced while the test set remains imbalanced (realistic)
```

LET'S TUNE THE MODEL USING HYPERPARAMETER TUNING, RandomizedSearchCV

```python
import joblib
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.linear_model import LogisticRegression
from sklearn.model_selection import RandomizedSearchCV
from sklearn.metrics import (
    confusion_matrix, accuracy_score, f1_score,
    precision_score, recall_score, roc_auc_score, classification_report
)
from scipy.stats import loguniform

# Define a broader parameter distribution for RandomizedSearch
param_dist = {
    'C': loguniform(1e-4, 1e2),        # log-uniform from 0.0001 to 100
    'penalty': ['l1', 'l2'],
    'solver': ['liblinear', 'saga'],
    'class_weight': [None, 'balanced'],
    'max_iter': [2000]  # slightly higher to ensure convergence with random params
}

# Initialize model
logreg = LogisticRegression()

# Use RandomizedSearchCV
print("Starting Randomized Hyperparameter Search...")
random_search = RandomizedSearchCV(
    estimator=logreg,
    param_distributions=param_dist,
    n_iter=50,               # number of sampled parameter settings
    cv=5,
    scoring='f1',            # optimize for F1
    n_jobs=-1,
    verbose=1,
    random_state=42,         # for reproducibility
    return_train_score=True
)

# Fit
random_search.fit(X_train, y_train)

# Best model
best_model = random_search.best_estimator_
print("\n✨ Best Parameters:", random_search.best_params_)

# --- Save best model to disk ---
model_path = "best_logistic_regression_model.joblib"
joblib.dump(best_model, model_path)
print(f"\n Best model saved to: {model_path}")

# --- Load model (example for future use) ---
# loaded_model = joblib.load(model_path)
    
# --- Get predictions ---
y_pred = best_model.predict(X_test)
y_pred_proba = best_model.predict_proba(X_test)[:, 1]  # for ROC-AUC

# --- Confusion Matrix ---
def ConfusionMatrix(y_true, y_pred):
    cm = confusion_matrix(y_true, y_pred)
    plt.figure(figsize=(6,4))
    sns.heatmap(cm, annot=True, fmt='d', cmap='Blues',
                xticklabels=['Predicted 0', 'Predicted 1'],
                yticklabels=['Actual 0', 'Actual 1'])
    plt.ylabel('Actual')
    plt.xlabel('Predicted')
    plt.title('Confusion Matrix (Best Model)')
    plt.show()
    return cm

cm = ConfusionMatrix(y_test, y_pred)

# --- Comprehensive Metrics ---
acc = accuracy_score(y_test, y_pred)
f1 = f1_score(y_test, y_pred)
precision = precision_score(y_test, y_pred)
recall = recall_score(y_test, y_pred)
roc_auc = roc_auc_score(y_test, y_pred_proba)

print("\n✅ Final Test Performance:")
print(f"Accuracy : {acc:.4f}")
print(f"Precision: {precision:.4f}")
print(f"Recall   : {recall:.4f}")
print(f"F1 Score : {f1:.4f}")
print(f"ROC-AUC  : {roc_auc:.4f}")

# Optional: Full classification report
print("\n📋 Detailed Classification Report:")
print(classification_report(y_test, y_pred, target_names=['Class 0', 'Class 1']))

# --- Top 5 Results from Random Search ---
cv_results = pd.DataFrame(random_search.cv_results_)
top5 = cv_results.nlargest(5, 'mean_test_score')  # higher F1 = better

top5['param_str'] = top5['params'].apply(
    lambda p: ', '.join(f"{k}={v}" for k, v in p.items())
)

print("\n🏆 Top 5 Random Search Trials (by F1):")
for i, (_, row) in enumerate(top5.iterrows()):
    print(f"{i+1}. F1 = {row['mean_test_score']:.4f} | {row['param_str']}")

# --- Plot Top 5 F1 Scores ---
plt.figure(figsize=(10, 5))
bars = plt.barh(range(len(top5)), top5['mean_test_score'], color='teal', edgecolor='black')
plt.yticks(range(len(top5)), [f"Trial {i+1}" for i in range(len(top5))])
plt.xlabel('Mean CV F1 Score')
plt.title('Top 5 Random Search Trials')
plt.gca().invert_yaxis()
for i, v in enumerate(top5['mean_test_score']):
    plt.text(v + 0.001, i, f"{v:.4f}", va='center', fontweight='bold')
plt.tight_layout()
plt.show()
```

```python
import joblib
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from xgboost import XGBClassifier
from sklearn.model_selection import RandomizedSearchCV
from sklearn.metrics import (
    confusion_matrix, accuracy_score, f1_score,
    precision_score, recall_score, roc_auc_score, classification_report
)
from scipy.stats import uniform, randint

# Define hyperparameter search space for XGBoost
param_dist = {
    'n_estimators': randint(100, 600),
    'max_depth': randint(3, 12),
    'learning_rate': uniform(0.01, 0.3),        # sampled from [0.01, 0.31)
    'subsample': uniform(0.6, 0.4),             # [0.6, 1.0)
    'colsample_bytree': uniform(0.6, 0.4),      # [0.6, 1.0)
    'gamma': uniform(0, 5),
    'reg_alpha': uniform(0, 10),                # L1 regularization
    'reg_lambda': uniform(1, 10),               # L2 regularization
    'min_child_weight': randint(1, 10),
    'scale_pos_weight': [1, 2, 5]               # useful for imbalanced data
}

# Initialize XGBoost model
xgb = XGBClassifier(
    use_label_encoder=False,
    eval_metric='logloss',
    random_state=42,
    n_jobs=-1
)

# Set up RandomizedSearchCV
print(" Starting Randomized Hyperparameter Search for XGBoost...")
random_search = RandomizedSearchCV(
    estimator=xgb,
    param_distributions=param_dist,
    n_iter=50,              # number of random combinations to try
    cv=5,
    scoring='f1',           # optimize for F1 (change to 'roc_auc' if preferred)
    n_jobs=-1,
    verbose=1,
    random_state=42
)

# Fit the search
random_search.fit(X_train, y_train)

# Best model
best_xgb = random_search.best_estimator_
print("\n Best Parameters:")
for k, v in random_search.best_params_.items():
    print(f"  {k}: {v}")

# Save model
model_path = "best_xgboost_model.joblib"
joblib.dump(best_xgb, model_path)
print(f"\n Best XGBoost model saved to: {model_path}")

# Predictions
y_pred = best_xgb.predict(X_test)
y_pred_proba = best_xgb.predict_proba(X_test)[:, 1]

# Confusion Matrix
def ConfusionMatrix(y_true, y_pred):
    cm = confusion_matrix(y_true, y_pred)
    plt.figure(figsize=(6, 4))
    sns.heatmap(cm, annot=True, fmt='d', cmap='Blues',
                xticklabels=['Predicted 0', 'Predicted 1'],
                yticklabels=['Actual 0', 'Actual 1'])
    plt.ylabel('Actual')
    plt.xlabel('Predicted')
    plt.title('Confusion Matrix (XGBoost)')
    plt.show()
    return cm

cm = ConfusionMatrix(y_test, y_pred)

# Metrics
acc = accuracy_score(y_test, y_pred)
f1 = f1_score(y_test, y_pred)
precision = precision_score(y_test, y_pred)
recall = recall_score(y_test, y_pred)
roc_auc = roc_auc_score(y_test, y_pred_proba)

print("\n Final Test Performance (XGBoost):")
print(f"Accuracy : {acc:.4f}")
print(f"Precision: {precision:.4f}")
print(f"Recall   : {recall:.4f}")
print(f"F1 Score : {f1:.4f}")
print(f"ROC-AUC  : {roc_auc:.4f}")

print("\n Classification Report:")
print(classification_report(y_test, y_pred, target_names=['Class 0', 'Class 1']))

# --- Optional: Top 5 Trials ---
cv_results = pd.DataFrame(random_search.cv_results_)
top5 = cv_results.nlargest(5, 'mean_test_score')
top5['param_str'] = top5['params'].apply(lambda p: ', '.join(f"{k}={v:.3g}" for k, v in p.items()))

print("\n Top 5 XGBoost Trials (by F1):")
for i, (_, row) in enumerate(top5.iterrows()):
    print(f"{i+1}. F1 = {row['mean_test_score']:.4f} | {row['param_str']}")
```
